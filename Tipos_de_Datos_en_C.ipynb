# Tipos de datos en C

Los tipos primitivos son las unidades básicas en C para reservar espacio en memoria RAM.

**char:** Guarda un solo carácter. Regla: Usa siempre comillas simples (ej. 'F').

**int:** Guarda números enteros positivos o negativos (sin decimales).

**float:** Guarda números reales con punto decimal (precisión estándar).

**double:** Guarda decimales con el doble de precisión que un float.

```python
%%script bash
cat << 'EOF' > primitivos.c

#include <stdio.h>
#include <stdbool.h>

void crearAtributosBasicos() {
    // Nombre de mi mascota: Python

    char inicialMascota = 'P';
    int edadDias = 5;
    float pesoKg = 2.35;
    double temperatura = 37.654;
    bool estaDormido = false;

    printf("Los atributos de mi mascota \n");
    printf("Inicial: %c\n", inicialMascota);
    printf("Edad: %d días\n", edadDias);
    printf("Peso: %.2f kg\n", pesoKg);
    printf("Temperatura: %.3lf °C\n", temperatura);
    if (estaDormido)
    {
      printf( "SI (Dormidito ...)");
    }
    else
    {
      printf(" NO (Super despierto!)");
    }

}

int main() {
    crearAtributosBasicos();
    return 0;
}
EOF

# Compilar y ejecutar
gcc primitivos.c -o primitivos
./primitivos
```

# Modificadores de Tipo (short, long, unsigned)

Los modificadores alteran el tamaño en memoria RAM y el rango numérico que puede guardar una variable.

**short** (2 bytes / 16 bits)

Rango: $-32,768$ a $32,767$

Uso: Guarda enteros pequeños para ahorrar espacio en memoria.


**unsigned short (2 bytes / 16 bits)**:
Rango: $0$ a $65,535$

Uso: Al eliminar los negativos, dobla la capacidad para números positivos pequeños.

**unsigned int (4 bytes / 32 bits):**

Rango: $0$ a $4,294,967,295$

Uso: Para contadores grandes donde es imposible tener valores negativos.

**long long (8 bytes / 64 bits):**

Rango: $-9\times 10^{18}$ a $9\times 10^{18}$ (números astronómicos)

Uso: Para identificar chips, contar pasos o latidos acumulados en la vida.

```python
%%script bash
cat << 'EOF' > modificadores.c

#include <stdio.h>

void demostrarModificadores() {
    // SHORT: Con signo (2 bytes)
    short nivelEnergia = 85;             // Porcentaje (0 a 100)
    short horasSueno = 12;               // Horas dormidas al día

    // USHORT: Sin signo (2 bytes)
    unsigned short galletasComidas = 450; // Conteo de premios recibidos
    unsigned short minutosJugados = 1200; // Minutos totales de juego

    // INT: Sin signo (4 bytes)
    unsigned int puntosFelicidad = 150000U; // Puntuación acumulada
    unsigned int idChipMascota = 3892011400U; // ID único de registro

    // LONG LONG: Enteros masivos (8 bytes)
    long long pasosTotales = 9876543210LL;    // Pasos acumulados en la vida
    long long latidosCorazon = 1250000000LL; // Latidos totales

    printf("*** ATRIBUTOS MODIFICADOS DE LA MASCOTA *** \n\n");

    printf("[SHORT - Con signo, 2 bytes]\n");
    printf("- Nivel de energía: %d%%\n", nivelEnergia);
    printf("- Horas de sueño: %d hrs\n\n", horasSueno);

    printf("[UNSIGNED SHORT - Sin signo, 2 bytes]\n");
    printf("- Galletas comidas: %u galletas\n", galletasComidas);
    printf("- Minutos jugados: %u min\n\n", minutosJugados);

    printf("[UNSIGNED INT - Sin signo, 4 bytes]\n");
    printf("- Puntos de felicidad: %u pts\n", puntosFelicidad);
    printf("- ID de Chip: %u\n\n", idChipMascota);

    printf("[LONG LONG - Enteros masivos, 8 bytes]\n");
    printf("- Pasos totales: %lld pasos\n", pasosTotales);
    printf("- Latidos del corazón: %lld latidos\n", latidosCorazon);
}

int main() {
    demostrarModificadores();
    return 0;
}
EOF

# Compilar y ejecutar
gcc modificadores.c -o modificadores
./modificadores
```

# Tipos Compuestos (enum, struct, union)

*Permiten agrupar variables o definir nuevos tipos de datos personalizados para representar entidades más complejas del mundo real.*

**enum (Enumeración):** Lista de opciones o constantes fijas asignadas a nombres legibles. En lugar de usar números sueltos (0, 1, 2), usa palabras como FELIZ, TRISTE, DORMIDO. Internamente, C les asigna valores enteros secuenciales.

**struct (Estructura):** Colección o carpeta que agrupa variables de diferentes tipos bajo un mismo nombre. Permite empaquetar toda la ficha de la mascota (char, int, float, bool, enum) en una sola variable contenedora.

**union (Unión):** Estructura donde todos sus miembros comparten exactamente el mismo espacio de memoria RAM. Su tamaño es igual al miembro más grande y solo puede guardar un dato activo a la vez (por ejemplo, el accesorio equipado de la mascota).

```python
%%script bash
cat << 'EOF' > enum_simple.c

#include <stdio.h>

// ENUM
enum EstadoAnimo {
    FELIZ,      // C le asigna internamente el valor 0
    TRISTE,     // C le asigna internamente el valor 1
    DORMIDO     // C le asigna internamente el valor 2
};

int main() {
    // Declaramos la variable usando el enum
    enum EstadoAnimo animoMascota = TRISTE;

    printf("** ESTADO DE ÁNIMO DE LA MASCOTA \n\n");

    // Mostramos el valor en texto según la opción seleccionada

    // Se puede hacer con switch case??

    if (animoMascota == FELIZ) {
        printf("La mascota está alegre y jugando. (Valor en memoria: %d)\n", animoMascota);
    } else if (animoMascota == TRISTE) {
        printf("La mascota necesita atención. (Valor en memoria: %d)\n", animoMascota);
    } else if (animoMascota == DORMIDO) {
        printf("La mascota está descansando. Zzz... (Valor en memoria: %d)\n", animoMascota);
    }

    return 0;
}
EOF

# Compilar y ejecutar
gcc enum_simple.c -o enum_simple
./enum_simple
```

```python
%%script bash
cat << 'EOF' > compuestos.c

#include <stdio.h>
#include <stdbool.h>

// ENUM: Estados de ánimo de la mascota
enum EstadoAnimo {
    FELIZ,      // Equivalente a 0
    TRISTE,     // Equivalente a 1
    HAMBRIENTO, // Equivalente a 2
    DORMIDO     // Equivalente a 3
};

// UNION: Compartir espacio para un accesorio (Gorra o Collar)
union Accesorio {
    int idGorra;        // Código numérico
    char letraCollar;   // Letra inicial tallada
};

// STRUCT: Ficha completa que agrupa todos los datos de la mascota
struct MascotaVirtual {
    char inicial;
    int edadDias;
    float pesoKg;
    bool tieneHambre;
    enum EstadoAnimo animoActual;
    union Accesorio accesorioEquipado;
};

void demostrarCompuestos() {
    // Creación e inicialización de la mascota usando la estructura
    struct MascotaVirtual miMascota;

    miMascota.inicial = 'F';
    miMascota.edadDias = 15;
    miMascota.pesoKg = 3.20f;
    miMascota.tieneHambre = false;
    miMascota.animoActual = FELIZ;
    miMascota.accesorioEquipado.idGorra = 104; // Asignamos ID de gorra

    printf("FICHA COMPLETA DE LA MASCOTA (STRUCT) \n\n");
    printf("- Inicial: %c\n", miMascota.inicial);
    printf("- Edad: %d días\n", miMascota.edadDias);
    printf("- Peso: %.2f kg\n", miMascota.pesoKg);
    printf("- ¿Tiene hambre?: %s\n", miMascota.tieneHambre ? "SÍ" : "NO");

    // Uso de ENUM
    printf("- Estado de animo [enum]: ");
    switch (miMascota.animoActual) {
        case FELIZ: printf("FELIZ (Código 0)\n"); break;
        case TRISTE: printf("TRISTE (Código 1)\n"); break;
        case HAMBRIENTO: printf("HAMBRIENTO (Código 2)\n"); break;
        case DORMIDO: printf("DORMIDO (Código 3)\n"); break;
    }

    // Uso de UNION
    printf("Accesorio equipado [union]: Correa ID #%d\n", miMascota.accesorioEquipado.idGorra);
}

int main() {
    demostrarCompuestos();
    return 0;
}
EOF

# Compilar y ejecutar
gcc compuestos.c -o compuestos
./compuestos
```
